```python

import pandas as pd
```

```python
#Loading data
url = "https://raw.githubusercontent.com/ritaafrica/data/refs/heads/main/student_scores.csv"
```

```python
df = pd.read_csv(url)
```

```python
df
```

```python
df.head()
```

```python
#Adding an new column for total
df["Total_Score"] = df["Math_Score"] + df["Reading_Score"] + df["Writing_Score"]
```

```python
df.head()
```

```python
#Creating a percentage score column
df["Percentage_Score"] = (df["Total_Score"] / 300) * 100
```

```python
df.head()
```

```python
df["Percentage_Score"] = df["Percentage_Score"].round(2)
```

```python
df.head()
```

```python
#Renaming column
df.rename(columns={"Math_Score": "Mathematics_Score"}, inplace=True)
```

```python
df.head()
```

```python
df.rename(columns={"Gender": "G"}, inplace=True)
```

```python
df.head()
```

```python
df.drop(columns=["School"], inplace=True)
```

```python
df.head()
```

```python
df.to_csv('modified_student_score.csv', index=False)
```

```python
df.to_html('modified_student_score.html', index=False)
```
